```python
import torch
from transformers import AutoImageProcessor, AutoModelForImageClassification
import os

repo_id = "codewithdark/vit-chest-xray"
local_dir = "/home/labo/Documents/ExAI/V3/Streamlit/saved_model/vit_chest_xray"

# Real labels from the model card (order matters!)
labels = ['Cardiomegaly', 'Edema', 'Consolidation', 'Pneumonia', 'No Finding']

processor = AutoImageProcessor.from_pretrained(repo_id)
model = AutoModelForImageClassification.from_pretrained(repo_id, torch_dtype=torch.float32)
model.eval()

# Patch config mappings
model.config.id2label = {i: lab for i, lab in enumerate(labels)}
model.config.label2id = {lab: i for i, lab in enumerate(labels)}

# Save locally (now config.json will contain your real label names)
os.makedirs(local_dir, exist_ok=True)
processor.save_pretrained(local_dir)
model.save_pretrained(local_dir)

print("Saved to:", local_dir)
print("id2label:", model.config.id2label)
```
